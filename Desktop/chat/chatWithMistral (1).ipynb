```python
import os
import json
from langchain_ollama import OllamaLLM
from langchain.prompts import PromptTemplate
from langchain.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.chains import RetrievalQA

# ✅ Charger les données d'un prospect spécifique à partir d'un fichier JSON
def load_prospect(file_path):
    with open(file_path, "r", encoding="utf-8") as file:
        data = json.load(file)
    return data  # Retourne un seul prospect

# ✅ Modèle d’embeddings pour indexer le prospect
embedding_model = HuggingFaceEmbeddings(model_name="BAAI/bge-small-en-v1.5")

# ✅ Séparer les textes en morceaux exploitables
text_splitter = RecursiveCharacterTextSplitter(chunk_size=800, chunk_overlap=200)

# ✅ Formater le texte du prospect pour l'analyser
def format_prospect(prospect):
    formatted_text = f"""
    📌 **Entreprise** : {prospect.get("company", "N/A")}
    🔹 **Secteur d'activité** : {prospect.get("activity", "N/A")}
    🔹 **Code NAF** : {prospect.get("code_naf", "N/A")}
    📧 **Email** : {prospect.get("email", "N/A")}
    📞 **Téléphone** : {prospect.get("phone", "N/A")}
    🎯 **Statut du lead** : {prospect.get("leadstatus", "Non spécifié")}
    💰 **Potentiel commercial** : {prospect.get("potentielprevisionnel", "Inconnu")}
    🏢 **Opérateur télécoms** : {prospect.get("parkoperator", "N/A")}
    📍 **Localisation** : {prospect.get("city", "N/A")}, {prospect.get("state", "N/A")}
    📅 **Dernier appel** : {prospect.get("call_date", "N/A")} à {prospect.get("call_time", "N/A")}
    📝 **Notes** : {prospect.get("description", "Aucune note disponible")}

    📌 **Informations supplémentaires** :
    - **Designation** : {prospect.get("designation", "N/A")}
    - **Email secondaire** : {prospect.get("secondaryemail", "Non fourni")}
    - **Téléphone standard** : {prospect.get("phonestd", "Non fourni")}
    - **Etablissement** : {prospect.get("establishment", "N/A")}
    - **SIREN** : {prospect.get("siren", "N/A")}
    - **SIRET** : {prospect.get("siret", "N/A")}
    - **Nombre d'employés** : {prospect.get("noofemployees", "Non spécifié")}
    - **Nombre de sites** : {prospect.get("noofsite", "Non spécifié")}
    - **Projet** : {prospect.get("project", "Non spécifié")}
    - **Qualification** : {prospect.get("qualifcategory", "Non spécifié")}
    - **Marque** : {prospect.get("market", "Non spécifié")}

    📌 **Détails du matériel** :
    - **Type de matériel** : {prospect.get("parkequipmenttype", "N/A")}
    - **Fournisseur PABX** : {prospect.get("park_equipment_provider_pabx", "Non précisé")}
    - **Type de connexion Internet** : {prospect.get("parkaccesstype", "Non précisé")}
    - **Débit Internet** : {prospect.get("park_internet_debit", "Non précisé")}
    - **Type de connexion mobile** : {prospect.get("park_mobile_nbr_lines", "Non précisé")}
    - **Opérateur Internet** : {prospect.get("parkoperatorinternet", "Non précisé")}

    📌 **État des équipements** :
    - **Maintenance des équipements** : {prospect.get("park_equipment_dfe_maintenance", "Non précisé")}
    - **Engagement des équipements** : {prospect.get("park_equipment_free_engagement", "Non précisé")}
    - **Maintenance SD-WAN** : {prospect.get("park_security_has_sdwan_peoject", "Non précisé")}

    📌 **Contact et suivi** :
    - **Agent de contact** : {prospect.get("meeting_agent", "Non précisé")}
    - **Commentaires du rendez-vous** : {prospect.get("meeting_comment", "Non précisé")}
    - **Date du rendez-vous** : {prospect.get("meeting_date", "Non précisé")}
    """
    return formatted_text


# ✅ Indexer le prospect dans ChromaDB
def index_prospect_in_chroma(prospect):
    formatted_text = format_prospect(prospect)
    split_texts = text_splitter.split_text(formatted_text)

    # Stocker les morceaux de texte et leurs métadonnées
    chunks = []
    metadata = [prospect]

    # Indexer dans ChromaDB
    vectorstore = Chroma.from_texts(split_texts, embedding=embedding_model, metadatas=metadata)
    retriever = vectorstore.as_retriever()

    print("✅ Base de données vectorielle ChromaDB créée avec succès pour ce prospect!")
    return retriever

# ✅ Charger le modèle Mistral via Ollama
llm = OllamaLLM(model="mistral")  # Utiliser le modèle Mistral ici

# ✅ Prompt avancé pour générer des analyses plus précises
analysis_prompt = PromptTemplate(
    input_variables=["context"],
    template="""
Voici les informations complètes d'un prospect :

{context}

**Analyse approfondie du prospect :**

1. **État actuel du prospect** :
   - Déterminez le statut actuel du prospect (Ex. : À rappeler, Intéressé, Refusé, etc.).

2. **Potentiel commercial** :
   - Évaluez le potentiel commercial du prospect (Ex. : Faible, Moyen, Élevé).

3. **Défis et obstacles à la conversion** :
   - Décrivez les défis ou obstacles qui empêchent la conversion de ce prospect. Identifiez les points de friction.

4. **Stratégie recommandée pour la conversion** :
   - Proposez une stratégie détaillée pour convertir ce prospect, en prenant en compte ses besoins, attentes et obstacles.

5. **Prochaines actions recommandées** :
   - Donnez des recommandations claires sur les prochaines étapes à suivre pour progresser dans la relation avec ce prospect.

6. **Génération dynamique des questions à poser au prospect** :
   - Sur la base de l'analyse ci-dessus, proposez des questions personnalisées que l'agent doit poser au prospect pour mieux comprendre ses besoins, attentes et processus décisionnel.
     Voici quelques exemples :

     1. **Quel est votre budget prévu pour les solutions de télécommunications ?**
     2. **Avez-vous déjà travaillé avec un fournisseur de services similaires ?**
     3. **Quels sont les critères les plus importants pour vous dans le choix d’un fournisseur ?**
     4. **Quelles fonctionnalités spécifiques recherchez-vous dans une solution ?**
     5. **Y a-t-il des défis spécifiques que vous rencontrez avec vos solutions actuelles ?**

7. **Questions potentielles que le prospect pourrait poser à l'agent** :
   - Sur la base des informations du prospect, proposez également des réponses recommandées pour les questions que le prospect pourrait poser à l'agent. Voici quelques exemples :

     1. **Quelles garanties de performance offrez-vous ?**
        - Réponse recommandée : "Nous offrons des garanties de performance adaptées aux solutions que nous proposons, avec des engagements de service clairs détaillés dans notre SLA."

     2. **Quels sont les coûts totaux et les options de financement ?**
        - Réponse recommandée : "Nous offrons plusieurs options de financement flexibles pour nous adapter à vos besoins. Un devis détaillé peut être fourni pour plus de transparence."

     3. **Comment assurez-vous la sécurité de mes données ?**
        - Réponse recommandée : "Nous utilisons des technologies de chiffrement avancées et mettons en œuvre des pratiques de sécurité rigoureuses pour protéger vos données contre toute menace."

     4. **Quelle est la durée de l'engagement pour vos solutions ?**
        - Réponse recommandée : "Nous proposons des solutions avec des engagements flexibles pour répondre à vos besoins spécifiques. Vous pouvez choisir la durée qui vous convient."

8. **Champs manquants pour compléter le profil du prospect** :
   - Identifiez les informations manquantes dans le profil du prospect pour compléter la fiche. Exemples :
     - **Téléphone** (si non renseigné)
     - **Statut du lead** (si non précisé)
     - **Budget estimé**
     - **Type de contrat ou de solution préférée**
     - **Équipements utilisés et leur état**
     - **Préférences d’opérateur ou fournisseur**
     - **Date de renouvellement des contrats en cours**

"""
)


# ✅ Fonction pour analyser un prospect avec Mistral
def analyze_prospect_with_mistral(prospect):
    # Indexer le prospect dans ChromaDB
    retriever = index_prospect_in_chroma(prospect)

    # Utiliser le retriever pour obtenir les informations du prospect
    retrieved_docs = retriever.invoke("Quel est l'état du prospect ?")  # Utiliser invoke() à la place de get_relevant_documents()

    if not retrieved_docs:
        return "❌ Aucun prospect trouvé."

    context = retrieved_docs[0].page_content

    # Générer la réponse en utilisant Mistral via le prompt
    response = llm.invoke(analysis_prompt.format(context=context))

    # Structurer la réponse dans des sous-champs
    analysis_result = {
        "etat": "",  # L'état actuel du prospect
        "potentiel": "",  # Le potentiel commercial du prospect
        "defis": "",  # Les défis ou obstacles
        "strategie": "",  # La stratégie recommandée
        "actions": "" , # Prochaines actions recommandées
        "questions_a_poser": "",  # Questions à poser au prospect
        "questions_du_prospect": "",  # Questions potentielles que le prospect pourrait poser
        "champs_manquants": "",  # Champs manquants
    }

    # Séparer les sections de la réponse de Mistral
    lines = response.strip().split("\n")
    if len(lines) >= 5:
        analysis_result["etat"] = lines[0].strip()  # L'état
        analysis_result["potentiel"] = lines[1].strip()  # Potentiel
        analysis_result["defis"] = lines[2].strip()  # Défis
        analysis_result["strategie"] = lines[3].strip()  # Stratégie recommandée
        analysis_result["actions"] = lines[4].strip()  # Actions recommandées
        analysis_result["questions_a_poser"] = lines[5].strip()  # Questions à poser au prospect
        analysis_result["questions_du_prospect"] = lines[6].strip()  # Questions du prospect
        analysis_result["champs_manquants"] = lines[7].strip()  # Champs manquants à collecter

    return analysis_result

# ✅ Sauvegarder l'analyse du prospect avec une bonne structuration
def save_analyzed_prospect(prospect, file_path):
    with open(file_path, "w", encoding="utf-8") as file:
        json.dump(prospect, file, indent=4, ensure_ascii=False)

# ✅ Exemple d'utilisation pour un seul prospect (par exemple "Adeline_JUMELLE_10x234.json")
prospect_file = "/content/your_extracted_folder/prospects_individuels/Adeline_JUMELLE_10x234.json"  # Spécifiez le chemin de votre prospect

# Charger le prospect à partir du fichier
prospect = load_prospect(prospect_file)

# Analyser le prospect avec Mistral
analyzed_prospect = analyze_prospect_with_mistral(prospect)

# Ajouter l'analyse structurée au prospect
prospect["analysis"] = analyzed_prospect

# Sauvegarder le prospect analysé dans un nouveau fichier
save_analyzed_prospect(prospect, f"/content/processed_{os.path.basename(prospect_file)}")

print(f"Analyse du prospect {prospect['company']} effectuée avec succès.")
```

```python
import json
import gradio as gr

# ✅ Charger les données des prospects
def load_prospect_data(json_file="prospects_information.json"):
    with open(json_file, "r", encoding="utf-8") as file:
        data = json.load(file)
    # ✅ Vérification de la structure
    if not isinstance(data, list):
        raise ValueError("❌ Erreur : Le fichier JSON ne contient pas une liste valide.")
    return data  # Retourne la liste des prospects

# Chargement des données de test ou du fichier JSON
prospects = load_prospect_data("prospects_information.json")

# ✅ Définir les champs obligatoires attendus
expected_fields = [
    "company", "activity", "code_naf", "email", "phone", "firstname", "lastname",
    "leadstatus", "potentielprevisionnel", "parkoperator", "city", "state",
    "call_date", "call_time", "description"
]

# ✅ Identifier les champs manquants pour chaque prospect et déterminer les champs critiques
def detect_missing_fields(prospects, expected_fields):
    missing_fields_list = []

    for prospect in prospects:
        missing_fields = [field for field in expected_fields if field not in prospect or not prospect[field]]

        # Champs critiques : ceux qui sont essentiels pour qualifier un prospect
        critical_fields = ["leadstatus", "phone", "email", "firstname"]

        missing_critical_fields = [field for field in critical_fields if field in missing_fields]

        missing_fields_list.append({
            "company": prospect.get("company", "N/A"),
            "missing_fields": missing_fields,
            "critical_missing_fields": missing_critical_fields,
        })

    # Trier par nom d'entreprise
    missing_fields_list.sort(key=lambda x: x["company"])

    return missing_fields_list

missing_fields_data = detect_missing_fields(prospects, expected_fields)

# ✅ Afficher uniquement les champs manquants pour chaque prospect
def display_missing_fields():
    result = []
    for entry in missing_fields_data:
        result.append(f"📌 **Entreprise**: {entry['company']}")
        result.append(f"   ❌ **Champs manquants**: {', '.join(entry['missing_fields'])}")
        result.append(f"   🔴 **Champs critiques manquants**: {', '.join(entry['critical_missing_fields'])}\n")
    return "\n".join(result)

# ✅ Utilisation de Gradio pour afficher le résultat de manière lisible
interface = gr.Interface(fn=display_missing_fields, inputs=[], outputs="text", title="Liste des Champs Manquants par Prospect")
interface.launch(share=True)
```

Attribue un score aux prospects

```python
import json
from langchain_ollama import OllamaLLM
from langchain.prompts import PromptTemplate
from langchain.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain.text_splitter import RecursiveCharacterTextSplitter

# Charger un prospect à partir d'un fichier JSON
def load_prospects(file_path):
    with open(file_path, "r", encoding="utf-8") as file:
        data = json.load(file)
    return data

# Formatage des données du prospect pour analyse
def format_prospect(prospect):
    formatted_text = f"""
    📌 **Entreprise** : {prospect.get("company", "N/A")}
    🔹 **Secteur d'activité** : {prospect.get("activity", "N/A")}
    📧 **Email** : {prospect.get("email", "N/A")}
    📞 **Téléphone** : {prospect.get("phone", "N/A")}
    🎯 **Statut du lead** : {prospect.get("leadstatus", "Non spécifié")}
    💰 **Potentiel commercial** : {prospect.get("potentielprevisionnel", "Inconnu")}
    """
    return formatted_text

# Attribuer un score à un prospect
def assign_score(prospect):
    score = 0
    # Exemple de critères pour l'attribution d'un score
    if prospect.get("potentielprevisionnel") == "BUSINESS":
        score += 10  # Scoring élevé pour les prospects avec un potentiel élevé
    if prospect.get("leadstatus") == "Intéressé":
        score += 5  # Score pour un prospect intéressé
    if prospect.get("activity") == "Commerce de gros (commerce interentreprises) de fournitures et équipements industriels divers":
        score += 8  # Secteur spécifique avec une note plus élevée
    if "rappelé" in prospect.get("description", "").lower():
        score += 2  # Ajouter des points pour les prospects qui ont été rappelés récemment
    return score

# Analyser et attribuer un score à chaque prospect
def prioritize_prospects(prospects):
    prioritized_prospects = []

    for prospect in prospects:
        score = assign_score(prospect)
        prospect["score"] = score  # Ajouter le score au prospect
        prioritized_prospects.append(prospect)

    # Trier les prospects par score décroissant (du plus élevé au plus bas)
    prioritized_prospects = sorted(prioritized_prospects, key=lambda x: x["score"], reverse=True)

    return prioritized_prospects

# Sauvegarder les prospects priorisés dans un fichier JSON
def save_prioritized_prospects(prioritized_prospects, output_file):
    with open(output_file, "w", encoding="utf-8") as file:
        json.dump(prioritized_prospects, file, indent=4, ensure_ascii=False)

# Charger les prospects à partir d'un fichier
prospects = load_prospects("prospects_information.json")  # Remplacez ce chemin par votre fichier

# Prioriser les prospects
prioritized_prospects = prioritize_prospects(prospects)

# Sauvegarder le fichier JSON des prospects priorisés
save_prioritized_prospects(prioritized_prospects, "/content/prioritized_prospects.json")

print("Les prospects ont été priorisés et sauvegardés avec succès dans 'prioritized_prospects.json'.")
```

```python
import json
import re

input_file = "dataset_prospect_analysis2 (3).jsonl"
output_file = "dataset_prospect_clean.jsonl"

# 🔧 Fonctions de nettoyage
def clean_text(text):
    if not isinstance(text, str):
        return ""
    return re.sub(r"\s+", " ", text.strip().replace("\n", " "))

def clean_email(email):
    return email.lower().strip() if email else ""

def title_case(text):
    return text.title().strip() if text else ""

def clean_fiche(fiche):
    champs_a_nettoyer = ["description", "other_comments", "meeting_comment", "lane"]
    for champ in champs_a_nettoyer:
        fiche[champ] = clean_text(fiche.get(champ, ""))

    fiche["email"] = clean_email(fiche.get("email", ""))
    fiche["secondaryemail"] = clean_email(fiche.get("secondaryemail", ""))
    fiche["city"] = title_case(fiche.get("city", ""))
    fiche["firstname"] = title_case(fiche.get("firstname", ""))
    fiche["lastname"] = title_case(fiche.get("lastname", ""))
    fiche["company"] = title_case(fiche.get("company", ""))
    fiche["activity"] = clean_text(fiche.get("activity", ""))
    return fiche

# 🔁 Lire, nettoyer et écrire
with open(input_file, "r", encoding="utf-8") as f:
    cleaned_data = []
    for line in f:
        try:
            item = json.loads(line)
            cleaned_item = item.copy()
            cleaned_item["input"] = clean_fiche(item.get("input", {}))
            cleaned_data.append({
                "instruction": item.get("instruction", "").strip(),
                "input": cleaned_item["input"],
                "output": item.get("output", "").strip()
            })
        except Exception as e:
            print("❌ Ligne ignorée à cause d'une erreur :", e)

# 💾 Sauvegarde
with open(output_file, "w", encoding="utf-8") as f_out:
    for item in cleaned_data:
        f_out.write(json.dumps(item, ensure_ascii=False) + "\n")

print("✅ Dataset nettoyé enregistré sous :", output_file)
```

```python
def validate_dataset(data):
    erreurs = []
    for idx, ex in enumerate(data):
        if not isinstance(ex, dict):
            erreurs.append((idx, "Entrée non valide (pas un dict)"))
            continue
        if not all(k in ex for k in ["instruction", "input", "output"]):
            erreurs.append((idx, "Clé manquante dans l'entrée"))
        if not isinstance(ex["input"], dict):
            erreurs.append((idx, "input n'est pas un dict"))
    return erreurs

# Exécuter
problems = validate_dataset(data)
if problems:
    print("❌ Problèmes trouvés :")
    for idx, err in problems:
        print(f"- Ligne {idx} → {err}")
else:
    print("✅ Toutes les entrées sont bien structurées !")
```
